```python

excel = pd.ExcelFile("Bank_Churn_Messy.xlsx")
print(excel.sheet_names)

```

```python

# Import both sheets
customer_df = pd.read_excel(
    "Bank_Churn_Messy.xlsx",
    sheet_name="Customer_Info"
)

account_df = pd.read_excel(
    "Bank_Churn_Messy.xlsx",
    sheet_name="Account_Info"
)

# Check their sizes
print("Customer table:", customer_df.shape)
print("Account table:", account_df.shape)

# Preview the first 5 rows
display(customer_df.head())
display(account_df.head())

# Check the column names
print("Customer columns:", customer_df.columns.tolist())
print("Account columns:", account_df.columns.tolist())
```

```python

# Join Account_Info to Customer_Info
df = pd.merge(
    account_df,
    customer_df,
    on="CustomerId",
    how="left",
    suffixes=("_account", "_customer")
)

print("Rows after join:", len(df))

# Count exact duplicate rows
duplicate_count = df.duplicated().sum()
print("Exact duplicate rows:", duplicate_count)

# Remove exact duplicate rows
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
print("Rows dropped:", duplicate_count)

# Check duplicate column names
print("Duplicate column names:",
      df.columns[df.columns.duplicated()].tolist())

display(df.head())
```

```python

# 1. Clean currency columns
for col in ["Balance", "EstimatedSalary"]:
    df[col] = (
        df[col].astype(str)
        .str.replace("€", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 2. Correct country names
df["Geography"] = df["Geography"].replace({
    "FRA": "France",
    "French": "France",
    "France": "France",
    "Spain": "Spain",
    "Germany": "Germany"
})

# 3. Check missing values
print("Missing values before filling:")
print(df.isnull().sum())

# 4. Fill missing values
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns

numeric_cols = df.select_dtypes(
    include="number"
).columns

for col in categorical_cols:
    df[col] = df[col].fillna("MISSING")

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# 5. Check numeric summaries
print("\nNumeric column summary:")
display(df.describe().round(2))

# 6. Check suspicious values
for col in ["CreditScore", "Age", "Balance",
            "EstimatedSalary", "NumOfProducts"]:
    print(f"\n{col}:")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())

print("\nMissing values after filling:")
print(df.isnull().sum().sum())
```

```python

# Replace the invalid salary with the median of valid salaries
valid_salary = df.loc[
    df["EstimatedSalary"] >= 0, "EstimatedSalary"
].median()

df.loc[
    df["EstimatedSalary"] < 0, "EstimatedSalary"
] = valid_salary

print("Corrected invalid salaries using median:", valid_salary)

# Check whether the two Tenure columns contain the same values
print(
    "Tenure columns identical:",
    df["Tenure_account"].equals(df["Tenure_customer"])
)

# Remove the redundant Tenure column if values match
if df["Tenure_account"].equals(df["Tenure_customer"]):
    df = df.drop(columns=["Tenure_customer"])
    df = df.rename(columns={"Tenure_account": "Tenure"})

print("Final columns:", df.columns.tolist())
print("Remaining missing values:", df.isnull().sum().sum())
```

```python

# Count customers who stayed and customers who left
churn_counts = df["Exited"].value_counts().sort_index()

print("Customers who stayed:", churn_counts.get(0, 0))
print("Customers who left:", churn_counts.get(1, 0))

# Create a bar chart
churn_counts.plot(kind="bar")

plt.title("Bank Customer Churn")
plt.xlabel("Exited (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```

```python

# Calculate churn percentage by country
country_churn = df.groupby("Geography")["Exited"].mean() * 100

print("Churn rate by country (%):")
print(country_churn.round(2))

# Plot country churn rates
country_churn.plot(kind="bar", title="Churn Rate by Country (%)")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()

# Calculate churn percentage by gender
gender_churn = df.groupby("Gender")["Exited"].mean() * 100

print("Churn rate by gender (%):")
print(gender_churn.round(2))

# Plot gender churn rates
gender_churn.plot(kind="bar", title="Churn Rate by Gender (%)")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python

numeric_features = [
    "CreditScore",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "EstimatedSalary"
]

for col in numeric_features:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=df, x="Exited", y=col)
    plt.title(f"{col} by Customer Churn")
    plt.xlabel("Exited (0 = Stayed, 1 = Left)")
    plt.ylabel(col)
    plt.show()
```

```python

numeric_features = [
    "CreditScore",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "EstimatedSalary"
]

for col in numeric_features:
    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=df,
        x=col,
        hue="Exited",
        bins=25,
        element="step"
    )

    plt.title(f"{col} Distribution by Churn Status")
    plt.xlabel(col)
    plt.ylabel("Number of Customers")
    plt.show()
```

```python

# Create a separate dataset for modeling
model_df = df.drop(
    columns=["CustomerId", "Surname"],
    errors="ignore"
).copy()

print("Original dataset shape:", df.shape)
print("Model dataset shape:", model_df.shape)

display(model_df.head())
```

```python

# Calculate balance relative to estimated salary
model_df["balance_v_income"] = (
    model_df["Balance"] /
    model_df["EstimatedSalary"].replace(0, np.nan)
)

# Handle any missing or infinite ratios
model_df["balance_v_income"] = (
    model_df["balance_v_income"]
    .replace([np.inf, -np.inf], np.nan)
)

model_df["balance_v_income"] = (
    model_df["balance_v_income"]
    .fillna(model_df["balance_v_income"].median())
)

# Display the new feature
print(model_df[["Balance", "EstimatedSalary",
                "balance_v_income", "Exited"]].head())

# Visualize the feature by churn status
sns.boxplot(data=model_df, x="Exited", y="balance_v_income")
plt.title("Balance-to-Income Ratio by Churn Status")
plt.xlabel("Exited (0 = Stayed, 1 = Left)")
plt.ylabel("Balance / Estimated Salary")
plt.show()
```

```python

# Convert categorical columns into dummy variables
model_df = pd.get_dummies(
    model_df,
    columns=[
        "Geography",
        "Gender",
        "HasCrCard",
        "IsActiveMember"
    ],
    drop_first=True,
    dtype=int
)

print("Final dataset shape:", model_df.shape)
print("\nFinal column names:")
print(model_df.columns.tolist())

display(model_df.head())
```

```python

# Save the final dataset
model_df.to_csv("bank_customer_model_ready.csv", index=False)

print("Project dataset saved successfully!")
print("Rows:", model_df.shape[0])
print("Columns:", model_df.shape[1])
print("Missing values:", model_df.isnull().sum().sum())

# Download the CSV to your computer
from google.colab import files
files.download("bank_customer_model_ready.csv")

```
